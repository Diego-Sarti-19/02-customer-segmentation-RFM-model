```python
import pandas as pd
import requests
import os
```

```python
url_origem = "https://raw.githubusercontent.com/daniel-usp/LIT/master/BancoDeDados.csv"

pasta_destino = r"C:\Users\diego\OneDrive\Documentos_OneDrive\02-Customer-classificationRFM\Dados\OlistEcommerce.csv"

# Carregar dados
df = pd.read_csv(url_origem)

# Salvar arquivo localmente
df.to_csv(
    pasta_destino,
    index=False,
    encoding="utf-8-sig"
)

print("Arquivo salvo com sucesso!")
```

```python
# URL de origem
url = "https://raw.githubusercontent.com/daniel-usp/LIT/master/BancoDeDados.csv"

# Criar diretório
os.makedirs("Dados", exist_ok=True)

# Baixar arquivo
response = requests.get(url, timeout=120)
response.raise_for_status()

# Salvar arquivo
with open("Dados/OlistEcommerce.csv", "wb") as f:
    f.write(response.content)

print("Download realizado com sucesso!")
print(f"Tamanho: {len(response.content) / (1024**2):.2f} MB")
```
